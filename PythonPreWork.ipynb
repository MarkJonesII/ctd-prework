```python
#This program will verify a users name, location, and age to enter a special event that is invite only.
#If the user RSVP and is in the correct location then they can enter.
#It will ask the user for their first name, where they're from, and how old they are.
#If verified against the system, then they can enter. Otherwise they are told to go home!
Username = input("Welcome to the Best Event of 2027! What is your first name? ")
UserLocation = input("Where are you from? ")
UserList = ("Bill", "Samantha", "Billy", "Chuck")
Location = ("New York", "Miami", "Boston", "Atlanta")
Age = input("How old are you? ")
AgeLimit = int(21)

if Username in UserList and UserLocation in Location and int(Age) >= AgeLimit:
  print ("Welcome " + Username + "! Enjoy the event!")
else:
  print ("Go home! Next!")


#Example run 1
#Input Bill, Miami, and 21
#Output: Welcome Bill! Enjoy the event!

#Example run 2
#Input Bob, New York, 18
#Output: Go home! Next!
```
